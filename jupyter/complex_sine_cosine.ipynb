# Euler

Euler's formula is crucial when applying complex numbers in digital signal processing, like in the Discrete Fourier Transform.
With *Euler's Number* $e \approx 2.71828$ - the base of the natural logarithm and exponential function - and the imaginary unit $j$, this complex exponential function is defined as:

$$ e^{j\varphi} = \cos(\varphi) + j \sin(\varphi)$$

The above equation states that we can use a combination of sine and cosine to express a complex number.
This is not only fundamental to DSP applications, but is also considered one of the *most fundamental equations in mathematics*...

# Visualization

We can visualize Euler's equation in the complex domain with the exponential representation of a complex number:

$$ z = r e^{j\varphi}$$

Real and imaginary part are the projections on the x-axes and the y-axes, respectively:


```python
fig = plt.figure()
ax = plt.subplot(2,1,1)

N = 200

# setup

ax.set_xlabel('Real',loc='right')
ax.set_ylabel('Imag',loc='top')

ax.set_xlim([-1.1,1.1])
ax.set_ylim([-1.1,1.1])
ax.set_aspect( 1 )

ax.spines['left'].set_position('center')
ax.spines['bottom'].set_position('center')
ax.spines['right'].set_color('none')
ax.spines['top'].set_color('none')

ax.xaxis.set_ticks([-1,0,1]) 
ax.yaxis.set_ticks([-1,0,1])

real=0
imag=0

z, = ax.plot([0,real],[0,imag],'k--', linewidth=2)
rea, = ax.plot([0,real],[0,0], 'k', linewidth=2)
ima, = ax.plot([0,0],[imag,0],'gray', linewidth=2)

x = np.linspace(0, 2 * np.pi, N)

ax2 = plt.subplot(2,1,2)
ax2.set_xlabel('$\\varphi$')
sin,  = ax2.plot(x, np.sin(x), color='gray', linewidth=2)
cos,  = ax2.plot(x, np.cos(x), color='k',linewidth=2)



def animate(s):

    phi = s/100 * np.pi;
    real = np.cos(phi)
    imag = np.sin(phi)

    rea.set_data([0,real],[0,0])
    ima.set_data([0,0],[imag,0])

    z.set_data([0,real],[0,imag]);
    sin.set_data(x[0:s], np.sin(x[0:s]))
    cos.set_data(x[0:s], np.cos(x[0:s]))


a = matplotlib.animation.FuncAnimation(fig, animate, frames=N);
plt.close()
from IPython.display import HTML
HTML(a.to_jshtml())
```

# Harmonic Functions with Complex Numbers

With the help of Euler, we can also express basic harmonic functions like sine and cosine through complex numbers:

$$ \cos{\varphi} = \frac{1}{2} (e^{j \varphi} + e^{-j \varphi})$$

$$ \sin{\varphi} = \frac{1}{2j} (e^{j \varphi} - e^{-j \varphi})$$

This relation is important when calculating the Fourier transform of specific signals.
