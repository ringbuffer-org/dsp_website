```python
fs          = 48000
L           = 50000

t = np.linspace(0,L/fs,L)
x = np.sin(2*np.pi*234*t)
phi = (2*np.pi*234*t)%(2*np.pi)

fig   = plt.figure()
   
ax    = fig.add_subplot(2, 1, 1)
line, = ax.plot(t[0:500],x[0:500])
ax.set_title('Sine Wave')
ax.set_xlabel("$t/s$");
ax.set_ylabel("$x(t)$");


ax2    = fig.add_subplot(2, 1, 2)
line, = ax2.plot(t[0:500],phi[0:500])
ax2.set_title('Wrapped Phase')
ax2.set_xlabel("$t/s$");
ax2.set_ylabel("$\\varphi(t)$");

fig.tight_layout() 



```
