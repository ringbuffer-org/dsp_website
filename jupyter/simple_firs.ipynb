```python
fs = 16000
N  = 16000

t   = np.linspace(0,1,N)
f   = np.linspace(0,fs,N)
h   = np.zeros_like(t) 

h[0]=-0.1
h[1]= 0.2
h[2]=-0.15

h=h/sum(h)

H = np.fft.fft(h)

plt.plot(f[0:int(N/2)],abs(H[0:int(N/2)]));
plt.xlabel('$f$/Hz');
plt.ylabel('$|H(f)|$');
```
