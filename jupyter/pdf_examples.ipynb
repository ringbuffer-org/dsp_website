```python
from scipy.fft import fft
import io
from urllib.request import urlopen

import numpy as np
import scipy as sp

from matplotlib import pyplot as plt
%config InlineBackend.figure_formats = ['svg']

import IPython.display as ipd

import warnings
warnings.filterwarnings('ignore')
from IPython.core.display import HTML
HTML("""
<style>
.jp-RenderedImage{
    display: table-cell;
    text-align: center;
    vertical-align: middle;
}
.jp-RenderedSVG{
    display: table-cell;
    text-align: center;
    vertical-align: middle;
}
</style>
""")
```

```python
# 1000 Hz sampling rate
fs = 48000

# time vector with N samples length
N  = 2*fs
t  = np.linspace(0,N,N)/fs

f0 = 200

Nsteps = 8

```

# Sine Wave

```python
x  = np.sin(2*np.pi*f0*t)

p,x = np.histogram(x,bins=100);
plt.plot(x[0:-1],p/sum(p),'gray')
plt.xlabel('$x$')
plt.ylabel('$p(x)$');
```

------

# Uniformly distributed Noise

```python
x=np.random.random(N)-0.5

p,x = np.histogram(x,bins=100);
plt.plot(x[0:-1],p/sum(p),'gray')
plt.xlabel('$x$')
plt.ylabel('$p(x)$');
plt.ylim([0,0.02])
ipd.Audio(x, rate=fs)
```

# Gaussian White Noise

```python
x=np.random.normal(0,0.1,N)

p,x = np.histogram(x,bins=100);
plt.plot(x[0:-1],p/sum(p),'gray')
plt.xlabel('$x$')
plt.ylabel('$p(x)$');
ipd.Audio(x, rate=fs)
```

# Speech

Noise and speech both show a Laplace distribution. The most frequent values have a low amplitude.

```python

url = 'https://ringbuffer.org/download/audio/voice/local_mushrooms.wav'
fs, xx = sp.io.wavfile.read(io.BytesIO(urlopen(url).read()))

p,x = np.histogram(xx,bins=100);
plt.plot(x[0:-1],p/sum(p),'gray')
plt.xlabel('$x$')
plt.ylabel('$p(x)$');

ipd.Audio(xx[:,1], rate=fs)

```
