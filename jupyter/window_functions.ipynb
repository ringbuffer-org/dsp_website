```python
import numpy as np
from scipy import signal
from scipy.fft import fft, fftshift
import matplotlib.pyplot as plt

```

Window functions have numerous application in digital signal processing.
They are used to truncate signals in the time-domain or spectra in the frequency-domain. 
This process is referred to as **windowing**.

No matter if windows are applied in the time- or frequency domain.

```python
window = signal.windows.boxcar(51)

plt.plot(window)

plt.title("Boxcar window")
plt.ylabel("Amplitude")
plt.xlabel("Sample")
```

```python
window = signal.windows.hamming(51)

plt.plot(window)

plt.title("Boxcar window")
plt.ylabel("Amplitude")
plt.xlabel("Sample")
```
