```python
# # Define the transfer function
num = [1, 0]  # Numerator coefficients
den = [1, -1, 0]  # Denominator coefficients
sys = control.TransferFunction(num, den)

# Get poles and zeros
poles, zeros = control.pzmap(sys, plot=False)

# Create the plot
plt.figure()
plt.plot(np.real(poles), np.imag(poles), 'rx', markersize=10)  # Poles
plt.plot(np.real(zeros), np.imag(zeros), 'bo', markersize=10)  # Zeros

# Add unit circle
theta = np.linspace(0, 2*np.pi, 100)
plt.plot(np.cos(theta), np.sin(theta), 'gray')

# Set axis limits and labels
plt.axis('equal')
plt.xlabel('Real')
plt.ylabel('Imaginary')
plt.title('Pole-Zero Plot')
plt.grid(True)
plt.show()
```
