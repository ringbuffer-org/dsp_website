```python
fs          = 48000
L           = 50000

 
t = np.linspace(0,L/fs,L)

x = np.sin(2*np.pi*234*t)

fig   = plt.figure()
ax    = fig.add_subplot(2, 1, 1)
line, = ax.plot(t[0:500],x[0:500])

plt.plot([0.001, 0.0053], [1.2,1.2],  '--', marker = '|', color='black')
plt.text(0.003, 0.8, "T", fontsize=12, color='black')

plt.plot([0, 0], [0,1], '--', marker = '_', color='black')
plt.text(0.0005, 0, "a", fontsize=12, color='black')

ax.set_title('Sine Wave')
ax.set_xlabel("$t/s$");
ax.set_ylabel("$x(t)$");

ipd.display(ipd.Audio(x, rate=fs)) 
```
